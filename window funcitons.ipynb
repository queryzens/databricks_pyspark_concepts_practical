```python
data = [
    ("IT", "Sneha", 120),
    ("IT", "Rohan", 100),
    ("IT", "Karan", 100),
    ("IT", "Neha", 80)
]

df = spark.createDataFrame(data, ["dept", "name", "score"])
display(df)
```

```python
display(df.groupBy("dept").agg({"score":"avg"}))
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# window spefiction ... department - high score
w = Window.partitionBy("dept").orderBy(F.col("score").desc())

# all window functions
display(df.withColumn("row_number", F.row_number().over(w)) \
    .withColumn("rank", F.rank().over(w)) \
        .withColumn("dense_rank", F.dense_rank().over(w)))
```

```python
df.withColumn("dense_rank", F.dense_rank().over(w)) \
    .filter(F.col("dense_rank") <=2) \
    .show()
```
